## Unadjusted Analysis

We draw inspiration from the [Tableone Vignette](https://pypi.org/project/tableone/).

```python
!pip3 install tableone
```

```python
from tableone import TableOne
```

```python
columns = ['d_score', 'v_score', 'j_score']
categorical = []
groupby = 'pass_fall'
nonnormal = ['d_score', 'v_score', 'j_score']
labels={'pass_fail': 'Passing VHHs'}
```

```python
df.columns
```

```python
mytable = TableOne(df,
                   columns=columns,
                   categorical=categorical,
                   groupby='pass_fail',
                   nonnormal=nonnormal,
                   rename=labels,
                   pval=False)
```

```python
print(mytable.tabulate(tablefmt = "fancy_grid"))
```

## Adjusted Analysis

We draw inspiration from Susan Li's [Building A Logistic Regression in Python, Step by Step](https://towardsdatascience.com/building-a-logistic-regression-in-python-step-by-step-becd4d56c9c8).

```python
import statsmodels.api as sm
```

```python
cols=['v_score', 'j_score', 'd_score']

X=df[cols]
y=df['pass_fail']
```

```python
df = df.dropna(subset=cols)
```

```python
df.shape
```

```python
logit_model=sm.Logit(y,X)
result=logit_model.fit()
print(result.summary2())
```
