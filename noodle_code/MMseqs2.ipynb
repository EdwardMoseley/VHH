To run linclust to generate a clustering of your database `DB` execute the following command. The result database follows the same format as the clustering format:

```python
!/home/linuxbrew/.linuxbrew/Cellar/mmseqs2/15-6f452/bin/mmseqs linclust DB DB_clu tmp
```

To extract the representative sequences from the clustering result call:

```python
!/home/linuxbrew/.linuxbrew/Cellar/mmseqs2/15-6f452/bin/mmseqs createsubdb DB_clu DB DB_clu_rep
```

```python
!/home/linuxbrew/.linuxbrew/Cellar/mmseqs2/15-6f452/bin/mmseqs convert2fasta DB_clu_rep DB_clu_rep.fasta
```
