<font size=5> Clonal Threshold Calculations (Scoper) </font>

Here we calculate clonal relatedness from the pairwise Hamming distance and automatically calculate the threshold to best separate the clonal populations.  Samples are separated into distinct clones


```r
library(scoper)
library(dplyr)
library(airr)
library(ggplot2)
library(shazam)

PhageDisplay <- list('10_S1','11_S2','12_S3','13_S4','14_S5','15_S6','16_S7','17_S8','18_S9','19_S10','20_S11','21_S12','22_S13','23_S14','24_S15','25_S16')


EVD68 <- list('A2A_S1','L2A_S2','PDA_S3','GblockA_S4','NegA_S5','A2G1_S6','L2G1_S7','PD1G1_S8','PD2G1_S9','GblockG1_S10','GloveG1_S11','A2G2_S12','L2G2_S13','PD1G2_S14','GblockG2_S15')
EVD68test <- list('A2G1_S6','L2G1_S7')
EVD68cat <- list('PhageDisplay','Alpaca','Gblock','Alex','Lara')

Small <- list("SSA-S1","SA-S1")
First <- list('A-S1','B-S3','C-S5','D-S7','E-S9','F-S11')
All <- list("A-S1",'B-S3','C-S5','D-S7','E-S9','F-S11','D-S2','A-S4','B-S6','E-S8','F-S10')
Last <- list('D-S2','A-S4','B-S6','E-S8','F-S10')

for (Sample in EVD68cat) {

#Sample <- "Sanger-masked"

WORK_DIR <- file.path("/home","troyhinkley","alpaca","EVD68","catout")
setwd(WORK_DIR)


SampleIN <- paste(Sample,"-RepOut.tsv",sep="")
SampleOUT <- paste(Sample,"-clones.tsv",sep="")
Sub <- "clones/"
PlotOUT <- paste(Sub,Sample,"-Clonal Threshold.pdf",sep="")

DB <- read_airr(file.path(SampleIN))

db1 <- distToNearest(DB, model="ham", normalize="len", 
                    vCallColumn="v_call", jCallColumn = "j_call",
                    sequenceColumn = "junction")

#threshold <- findThreshold(db1$dist_nearest, method = "density")
threshold <- findThreshold(db1$dist_nearest, method = "gmm", model="gamma-norm", cutoff="intersect")
thr <- round(threshold@threshold, 3)

results <- hierarchicalClones(DB, method="nt", threshold=thr)

results1 <- as.data.frame(results)
#glimpse(results1)
plotCloneSummary(results, binwidth=0.02, title=paste(Sample,"Clonal Threshold",thr))

ggsave(file=PlotOUT)

write_airr(results1, file.path(Sub,SampleOUT))


}

```

```r
library(scoper)
library(dplyr)
library(airr)
library(ggplot2)
library(shazam)

PhageDisplay <- list('11_S2','12_S3','13_S4','14_S5','15_S6','16_S7','17_S8','18_S9','19_S10','20_S11','21_S12','22_S13','23_S14','24_S15','25_S16')

EVD68 <- list('A2A_S1','L2A_S2','PDA_S3','GblockA_S4','A2G1_S6','L2G1_S7','PD1G1_S8','PD2G1_S9','GblockG1_S10','GloveG1_S11','A2G2_S12','L2G2_S13','PD1G2_S14','GblockG2_S15')
EVD68test <- list('A2G1_S6','L2G1_S7')
EVD68cat <- list('PhageDisplay','Alpaca','Gblock','Alex','Lara')
Small <- list("SSA-S1","SA-S1")
First <- list('A-S1','B-S3','C-S5','D-S7','E-S9','F-S11')
All <- list("A-S1",'B-S3','C-S5','D-S7','E-S9','F-S11','D-S2','A-S4','B-S6','E-S8','F-S10')
Last <- list('D-S2','A-S4','B-S6','E-S8','F-S10')

for (Sample in EVD68cat) {

#Sample <- "11_S2"

WORK_DIR <- file.path("/home","troyhinkley","alpaca","EVD68")
setwd(WORK_DIR)
Sub0 <- "catout/"
Sub1 <- "spectralclones/"

SampleIN <- paste(Sub0,Sample,"-RepOut.tsv",sep="")
SampleOUT <- paste(Sub1,Sample,"-clones.tsv",sep="")
LogOUT <- paste(Sub1,Sample,"-log.txt",sep="")
PlotOUT <- paste(Sub1,Sample,"-Clonal Threshold.pdf",sep="")

DB <- read_airr(file.path(SampleIN))

results <- spectralClones(DB, method="vj", nproc=16, cdr3=TRUE, iter_max=200, nstart = 50)

results1 <- as.data.frame(results)
#glimpse(results1)
plotCloneSummary(results, binwidth=0.02, title=paste(Sample,"Clonal Threshold"))

ggsave(file=PlotOUT)

write_airr(results1, file.path(SampleOUT))

}
```
